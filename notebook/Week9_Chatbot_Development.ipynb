```python

# Load chatbot responses
responses = pd.read_csv(
    dataset_dir / "responses.csv",
    encoding="utf-8-sig"
)

print("CAMPUSAI - CHATBOT RESPONSE ANALYSIS")
print("Total Responses:", len(responses))

# Display question, response and status
display(
    responses[
        [
            "Question_ID",
            "Category",
            "Question",
            "Response",
            "Status"
        ]
    ]
)
```

```python

# CampusAI chatbot performance analysis

total_questions = len(responses)

successful = (
    responses["Status"].str.strip().str.lower() == "success"
).sum()

failed = total_questions - successful

success_rate = (
    successful / total_questions * 100
    if total_questions > 0 else 0
)

average_time = responses["Response_Time_Seconds"].mean()

print("CAMPUSAI CHATBOT - PERFORMANCE REPORT")
print("------------------------------------")
print("Total Questions:", total_questions)
print("Successful Responses:", successful)
print("Failed Responses:", failed)
print(f"Success Rate: {success_rate:.2f}%")
print(f"Average Response Time: {average_time:.2f} seconds")

print("\nCATEGORY-WISE PERFORMANCE")

category_report = responses.groupby("Category").agg(
    Total_Questions=("Question_ID", "count"),
    Average_Response_Time=("Response_Time_Seconds", "mean")
)

display(category_report)
```


## Chatbot Observations and Analysis

### 1. Testing Methodology
The CampusAI chatbot was developed using Python and Ollama with the Gemma 3 (1B) model.

The chatbot was tested using 20 education-related questions across four categories:
- Factual questions
- Reasoning questions
- Coding questions
- General conversation

Responses were recorded in responses.csv along with timestamps, execution times and status information.

### 2. Strengths
- Runs locally using Ollama.
- Supports interactive conversations through Python.
- Maintains conversation history during a chatbot session.
- Can explain educational concepts and generate Python code.
- Automatically saves conversations in CSV format.
- Records timestamps and chatbot activity for later analysis.

### 3. Limitations
- Response generation depends on available computer resources.
- The model may generate incorrect or incomplete information.
- Long conversations may exceed the model's context window.
- Conversation history is maintained in memory during execution but is not automatically restored when the program restarts.
- The chatbot does not verify answers against external educational sources.

### 4. Prompt Improvements
The chatbot uses a system prompt instructing the model to provide clear educational explanations and simple examples.

Possible improvements include:
- Adding more specific instructions for coding questions.
- Asking the model to explain reasoning step by step.
- Limiting response length for faster generation.
- Adding a mechanism to check answers against reliable sources.

### 5. Future Scope
The project can be extended using LangChain to:
- Manage conversation memory.
- Create reusable prompt templates.
- Integrate educational documents.
- Implement retrieval-augmented generation (RAG).

### 6. Conclusion
CampusAI demonstrates how a local LLM can be used to build an interactive educational chatbot.

The project implements command-line interaction, conversation history, CSV logging and automated testing. It provides a foundation for future LangChain integration.

```python

import ollama
import pandas as pd
from datetime import datetime

model = "gemma3:1b"

prompts = {
    "Original": "Explain overfitting in Machine Learning.",
    "Improved": (
        "Explain overfitting in Machine Learning. "
        "Include a simple definition, a real-world example, "
        "its consequences, and two prevention techniques. "
        "Use clear headings and student-friendly language."
    )
}

results = []

for prompt_type, prompt in prompts.items():
    print(f"\nTesting {prompt_type} Prompt...")

    response = ollama.chat(
        model=model,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        options={
            "num_predict": 300,
            "num_ctx": 2048
        }
    )

    answer = response["message"]["content"]

    results.append({
        "Timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "Prompt_Type": prompt_type,
        "Prompt": prompt,
        "Response": answer
    })

    print("Response:")
    print(answer)
    print("-" * 60)

comparison = pd.DataFrame(results)

comparison.to_csv(
    dataset_dir / "prompt_comparison.csv",
    index=False
)

print("\nPrompt comparison saved successfully!")
display(comparison)
```
